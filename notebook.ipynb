## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import seaborn as sns
```

## Leitura dos dados

```python
df = pd.read_csv('./sechselaeuten.csv')
df.head()
```

```python
df.info()
```

## Qual a correlação entre as variáveis numéricas?

```python
# Transformando a coluna record em inteiro
df['record'] = df['record'].astype(int)
df.info()
```

```python
varivaeis_numericas = df.select_dtypes(include=['number']).columns
matriz_correlacao = df[varivaeis_numericas].corr()

# Criando um mapa de calor da matriz de correlação
plt.figure(figsize=(12, 8))
sns.heatmap(matriz_correlacao, annot=True, cmap='coolwarm', vmin=-1, vmax=1,fmt='.2f')
plt.title('Mapa de Calor da Matriz de Correlação')
plt.show()


```
